```python
import numpy as np
import matplotlib.pyplot as plt

# Function to separate the blood pressure readings ("systolic/diastolic")
def parse_bp(bp_str):
    systolic, diastolic = bp_str.split('/')
    return int(systolic), int(diastolic)

# Function to plot the averages over time
def plot_bp_data(data, title):
    time_points = len(data[0])

    # Create lists for systolic and diastolic values
    data_systolic = [[] for _ in range(time_points)]
    data_diastolic = [[] for _ in range(time_points)]

    # Populate the lists by parsing each blood pressure reading
    for row in data:
        for j in range(time_points):
            sys_val, dia_val = parse_bp(row[j])
            data_systolic[j].append(sys_val)
            data_diastolic[j].append(dia_val)

    # Calculate the averages for each time point
    avg_systolic = [np.mean(values) for values in data_systolic]
    avg_diastolic = [np.mean(values) for values in data_diastolic]

    time_labels = ['0 min', '15 mins', '30 mins', '45 mins', '60 mins']

    print(title)
    print("Average Systolic:", avg_systolic)
    print("Average Diastolic:", avg_diastolic)

    plt.figure(figsize=(8, 5))
    plt.plot(time_labels, avg_systolic, marker='o', label='Average Systolic')
    plt.plot(time_labels, avg_diastolic, marker='o', label='Average Diastolic')
    plt.xlabel('Time')
    plt.ylabel('Blood Pressure (mmHg)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

# Natural Caffeine Dataset
data1 = [
    ['107/62', '85/63', '97/64', '86/63', '97/69'],
    ['104/67', '95/78', '126/76', '117/76', '121/87'],
    ['91/60', '98/64', '104/72', '108/65', '97/74'],
    ['96/62', '97/80', '102/67', '106/68', '102/64'],
    ['125/84', '124/69', '124/74', '93/66', '122/74']
]

# Placebo Dataset
data2 = [
    ['137/91', '130/83', '131/91', '146/101', '151/93'],
    ['87/59',  '91/54',  '85/55',  '165/62', '161/64'],
    ['103/63', '106/79', '129/40', '98/67',  '106/65'],
    ['109/72', '125/98', '116/83', '106/89', '117/79'],
    ['121/81', '124/90', '146/80', '133/77', '129/83']
]

# Synthetic Caffeine Dataset
data3 = [
    ['96/60',  '101/64', '92/50',  '91/60',  '110/67'],
    ['107/78', '117/81', '129/75', '116/98', '129/97'],
    ['129/81', '137/100','134/93', '138/76', '137/79'],
    ['130/77', '122/86', '133/80', '144/74', '121/76'],
    ['131/86', '129/88', '121/78', '127/81', '144/87']
]

plot_bp_data(data1, "Natural Caffeine: Average Blood Pressure Over Time")
plot_bp_data(data2, "Placebo: Average Blood Pressure Over Time")
plot_bp_data(data3, "Synthetic Caffeine: Average Blood Pressure Over Time")
```

```python
# Install pingouin if not already installed
!pip install pingouin --quiet

# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pingouin as pg

# Helper function to parse a blood pressure reading (e.g., "107/62")
def parse_bp(bp_str):
    systolic, diastolic = bp_str.split('/')
    return int(systolic), int(diastolic)

# Define the three datasets:
# Dataset 1: Natural caffeine
data_natural = [
    ['107/62', '85/63',  '97/64',  '86/63',  '97/69'],
    ['104/67', '95/78',  '126/76', '117/76', '121/87'],
    ['91/60',  '98/64',  '104/72', '108/65', '97/74'],
    ['96/62',  '97/80',  '102/67', '106/68', '102/64'],
    ['125/84', '124/69', '124/74', '93/66',  '122/74']
]

# Dataset 2: Placebo
data_placebo = [
    ['137/91', '130/83', '131/91', '146/101', '151/93'],
    ['87/59',  '91/54',  '85/55',  '165/62',  '161/64'],
    ['103/63', '106/79', '129/40', '98/67',   '106/65'],
    ['109/72', '125/98', '116/83', '106/89',  '117/79'],
    ['121/81', '124/90', '146/80', '133/77',  '129/83']
]

# Dataset 3: Synthetic caffeine
data_synthetic = [
    ['96/60',  '101/64', '92/50',  '91/60',  '110/67'],
    ['107/78', '117/81', '129/75', '116/98', '129/97'],
    ['129/81', '137/100','134/93', '138/76', '137/79'],
    ['130/77', '122/86', '133/80', '144/74', '121/76'],
    ['131/86', '129/88', '121/78', '127/81', '144/87']
]

# The time points (in minutes)
time_points = [0, 15, 30, 45, 60]

# Create a long-format DataFrame.
# Since subjects are different across conditions, we generate unique subject IDs.
rows = []

# Mapping from condition names to their datasets.
conditions = {
    'Natural': data_natural,
    'Placebo': data_placebo,
    'Synthetic': data_synthetic
}

# Build the long-format data.
for cond, data in conditions.items():
    for subj_idx, subject_data in enumerate(data, start=1):
        subject_id = f"{cond}_{subj_idx}"  # Unique subject identifier
        for time_idx, time_val in enumerate(time_points):
            bp_str = subject_data[time_idx]
            systolic, diastolic = parse_bp(bp_str)  # For the analysis, we use systolic only.
            rows.append({
                'subject': subject_id,  # Unique subject id
                'condition': cond,      # 'Natural', 'Placebo', or 'Synthetic'
                'time': time_val,       # Time in minutes (0, 15, 30, 45, 60)
                'systolic': systolic    # Systolic blood pressure measurement
            })

# Create the DataFrame.
df = pd.DataFrame(rows)
print("Long-format DataFrame for Mixed ANOVA:")
print(df.head(10))

# ---------------------------
# Perform Mixed ANOVA
# ---------------------------
# In this design the 'condition' (drink) is a between–subjects factor and 'time' is a within–subjects factor.
anova_results = pg.mixed_anova(dv='systolic', within='time', between='condition', subject='subject', data=df)
print("\nMixed ANOVA Results:")
print(anova_results)

# ---------------------------
# Plot the Average Systolic Blood Pressure Over Time for Each Condition
# ---------------------------
mean_df = df.groupby(['condition','time'])['systolic'].mean().reset_index()

plt.figure(figsize=(8, 5))
for cond in df['condition'].unique():
    subset = mean_df[mean_df['condition'] == cond]
    plt.plot(subset['time'], subset['systolic'], marker='o', label=cond)
plt.xlabel('Time (mins)')
plt.ylabel('Average Systolic Blood Pressure (mmHg)')
plt.title('Mean Systolic Blood Pressure Over Time by Condition')
plt.legend(title='Condition')
plt.grid(True)
plt.show()

# ---------------------------
# Summary and Conclusion
# ---------------------------
# Compute overall average systolic blood pressure by condition.
condition_avg = df.groupby('condition')['systolic'].mean().reset_index()
print("\nOverall Average Systolic Blood Pressure by Condition:")
print(condition_avg)

# Identify which condition had the highest and lowest overall averages.
highest = condition_avg.loc[condition_avg['systolic'].idxmax()]
lowest  = condition_avg.loc[condition_avg['systolic'].idxmin()]

print(f"\nSummary:")
print(f"- The condition with the highest average systolic blood pressure is: {highest['condition']} ({highest['systolic']:.1f} mmHg)")
print(f"- The condition with the lowest average systolic blood pressure is: {lowest['condition']} ({lowest['systolic']:.1f} mmHg)")

print("\nConclusion:")
print("Based on the overall averages calculated from the data:")
if highest['condition'] != lowest['condition']:
    print(f"• The '{lowest['condition']}' group (Natural caffeine) exhibited the lowest average systolic blood pressure, while the '{highest['condition']}' group ({highest['condition']}) showed the highest average.")
    print("• Although the mixed ANOVA did not yield statistically significant differences at the conventional 0.05 level, the effect size for the condition factor was moderate.")
    print("• These results suggest that, under the current conditions and sample sizes, Natural caffeine may be associated with lower systolic blood pressure compared to Placebo and Synthetic caffeine.")
    print("• Further research with larger samples may help to confirm these trends and clarify the practical implications.")
else:
    print("There are no notable differences in average systolic blood pressure between the drinks based on the current analysis.")
```
